Importing Libraries

```python
import pandas as pd
import numpy as np
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
```

Task 1

```python
# load dataset
df = sns.load_dataset("titanic")
print("First 10 rows of the Titanic dataset:")
display(df.head(10))

# check missing values
print("Missing values in each column:")
print(df.isnull().sum())

# fill missing age with mean
mean_age = df["age"].mean()
df["age"] = df["age"].fillna(mean_age)
print("Missing Age values after filling:", df["age"].isnull().sum())

# drop rows where embarked is missing
df = df.dropna(subset=["embarked"])
print("Missing Embarked values after dropping:", df["embarked"].isnull().sum())
```

Task 2

```python
# Convert Sex Column into numerical values
df["sex"] = df["sex"].map({
    "male": 0,
    "female": 1
})

print("Encoded Sex:")
print(df["sex"].head(10))

# One-Hot Encoding for Embarked
df = pd.get_dummies(
    df,
    columns=["embarked"],
    prefix="Embarked",
    dtype=int
)

print("One-Hot Encoded Embarked:")
print(df[[
    "Embarked_C",
    "Embarked_Q",
    "Embarked_S"
]].head(10))
```

Task 3

```python
X = df[["age", "fare", "sex", "pclass"]]
print("Selected Features:")
display(X.head(10))

# Target variable
y = df["survived"]
print("Target:")
display(y.head(10))

# StandardScaler Scaling
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Convert back to DataFrame for easier viewing
X_scaled = pd.DataFrame(
    X_scaled,
    columns=["Age", "Fare", "Sex", "Pclass"]
)
print("Scaled Features:")
display(X_scaled.head(10))

# Split dataset
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled,
    y,
    test_size=0.20,
    random_state=42
)

print("Dataset Split Results:")
print("Training features:", X_train.shape)
print("Testing features :", X_test.shape)
print("Training target  :", y_train.shape)
print("Testing target   :", y_test.shape)

print("\nTraining Data:")
display(X_train.head())
print("\nTesting Data:")
display(X_test.head())
```
